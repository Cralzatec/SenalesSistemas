```python
%%writefile 1_Conceptos_Importantes.py
import streamlit as st

# ======================
# CONFIGURACIÓN ESTÉTICA
# ======================
COLOR_SCHEME = {
    'primary': '#E53935',
    'secondary': '#EF9A9A',
    'background': '#FFFFFF',
    'text': '#000000',
    'accent': '#D32F2F',
    'highlight': '#FFCDD2'
}

st.set_page_config(
    page_title="Conceptos Importantes",
    layout="wide",
    page_icon="📓"
)

# CSS Personalizado
st.markdown(f"""
<style>
    .stApp {{
        background-color: {COLOR_SCHEME['background']};
        color: {COLOR_SCHEME['text']};
        font-family: 'Segoe UI', sans-serif;
    }}
    h1, h2, h3, h4 {{
        color: {COLOR_SCHEME['primary']} !important;
        border-bottom: 2px solid {COLOR_SCHEME['secondary']};
        padding-bottom: 8px;
    }}
    .stMarkdown > hr {{
        border-color: {COLOR_SCHEME['secondary']};
        margin: 2rem 0;
    }}
    .stMarkdown {{
        line-height: 1.6;
        color: {COLOR_SCHEME['text']} !important;
    }}
    .katex {{
        font-size: 1.1em !important;
        color: {COLOR_SCHEME['accent']} !important;
    }}
    ul {{
        padding-left: 1.2rem;
    }}
    li {{
        margin-bottom: 0.5rem;
    }}
</style>
""", unsafe_allow_html=True)

# ======================
# CONTENIDO PRINCIPAL
# ======================
st.title("Conceptos Clave")
st.markdown("""
Aquí se enseñan las ideas básicas que hacen posible que podamos enviar mensajes, videos o música sin usar cables, como con el Wi-Fi o el celular. Se empieza con una herramienta matemática llamada Transformada de Fourier, que ayuda a entender las señales, y se llega hasta métodos más modernos como OFDM, que hacen que la señal sea más rápida y clara.
""")
st.markdown("---")

# Crear dos columnas
col1, col2 = st.columns(2)

with col1:

    st.subheader("Transformada de Fourier (FT, DFT, FFT)")
    st.markdown(r"""
    * **Definición:** Es una herramienta matemática que toma una señal que cambia con el tiempo y la divide en ondas más simples, mostrando qué frecuencias la forman. Esto ayuda a ver qué tanto de cada sonido o frecuencia hay dentro de la señal.
    * **Modelado Matemático (DFT):**
        $$X[k] = \sum_{n=0}^{N-1} x[n]e^{-j\frac{2\pi k n}{N}}$$
    * **Usos en Comunicaciones Inalámbricas:** .
    Es esencial para examinar el contenido espectral de una señal y para el desarrollo de moduladores y demoduladores. En tecnologías OFDM, facilita la conversión entre los dominios temporal y frecuencial, permitiendo transmitir múltiples subportadoras de manera simultánea y eficiente.
    * **Fuente:**
        * [Transformada Rápida de Fourier (FFT)](https://mathworld.wolfram.com/FastFourierTransform.html)
    """)

    st.markdown("---")


    st.subheader("Señales Analíticas y la Transformada de Hilbert")
    st.markdown(r"""
    * **Definición:** Representación compleja de una señal real.
    * **Modelado Matemático (Señal Analítica):**
        $$x_a(t) = x(t) + j\hat{x}(t)$$
    * **Usos en Comunicaciones Inalámbricas:**
     *   **Filtrado de Canal:** Seleccionar bandas de frecuencia y rechazar interferencias.
    *   **Anti-aliasing y Reconstrucción:** Prevenir el aliasing en ADCs y reconstruir señales en DACs.
    *   **Conformación de Pulso:** Reducir la interferencia intersímbolo y controlar el ancho de banda.
    *   **Reducción de Ruido:** Mejorar la relación señal/ruido.
    * **Fuente:**
        * [Transformada de Hilbert](https://es.wikipedia.org/wiki/Transformada_de_Hilbert)
    """)

    st.markdown("---")


    st.subheader("OFDM (Multiplexación por División de Frecuencias Ortogonales)")
    st.markdown(r"""
    * **Definición:** Técnica de modulación digital.
    * **Modelado Matemático:**
        $$x_n = \frac{1}{\sqrt{N}} \sum_{k=0}^{N-1} X_k e^{j\frac{2\pi nk}{N}}$$
    * **Usos en Comunicaciones Inalámbricas:**
    Es crucial para convertir señales moduladas de banda pasante en su forma de banda base compleja, lo que permite procesar de manera eficiente las componentes I y Q, y facilita la implementación de modulaciones como SSB.
    * **Fuente:**
        * [OFDM en Wikipedia](https://es.wikipedia.org/wiki/Multiplexaci%C3%B3n_por_divisi%C3%B3n_de_frecuencias_ortogonales)
    """)

with col2:

    st.subheader("Filtrado Digital (FIR / IIR)")
    st.markdown(r"""
    * **Definición:** Algoritmos que modifican el espectro de una señal digital.
    * **Modelado Matemático (Ejemplo FIR):**
        $$y[n] = \sum_{k=0}^{M} b_k x[n-k]$$
    * **Usos en Comunicaciones Inalámbricas:** .
    Ofrece una alta eficiencia en el uso del espectro y permite transmitir grandes cantidades de datos al enviar varios bits por cada símbolo. Es fundamental en tecnologías como WiMAX, LTE, Wi-Fi (802.11ac/ax) y 5G. Los diagramas de constelación muestran los puntos I/Q, donde cada punto corresponde a un símbolo específico.
    * **Fuente:**
        * [Filtro digital](https://es.wikipedia.org/wiki/Filtro_digital)
    """)

    st.markdown("---")


    st.subheader("Señales I/Q y Modulación QAM")
    st.markdown(r"""
    * **Definición:**
        * **Señales I/Q:** Representación como componentes en fase y en cuadratura.
        * **QAM:** Modula amplitud y fase.
    * **Modelado Matemático:**
        $$s(t) = I(t)\cos(2\pi f_c t) - Q(t)\sin(2\pi f_c t)$$
    * **Usos en Comunicaciones Inalámbricas:** .
     *   **Resistencia al Multitrayecto:** Las subportadoras de baja velocidad son menos susceptibles a los retardos.
    *   **Eficiencia Espectral:** La ortogonalidad permite la superposición espectral sin interferencia.
    *   **Flexibilidad:** Permite la asignación adaptativa de modulaciones a subportadoras según las condiciones del canal.
    * **Fuente:**
        * [Ultimate Guide to QAM](https://www.numberanalytics.com/blog/ultimate-guide-to-qam)
    """)

    st.markdown("---")


    st.subheader("Diagonalización del Canal mediante FFT")
    st.markdown(r"""
    #### Definición
    En sistemas OFDM.

    $r[n] = h[n] * x[n] + w[n]$

    Con Prefijo Cíclico:

    $R[k] = H[k] \cdot X[k] + W[k]$

    ####  Implicaciones
    - Simplifica la ecualización
    - Permite FFT/IFFT

    ####  Fuente
    - *A Short Introduction to OFDM*, Mérouane Debbah
    """)

    st.markdown("---")
```

###Dominio de la Frecuencia

```python
%%writefile 2_Dominio_de_la_Frecuencia.py
import streamlit as st
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import butter, lfilter, freqz
from matplotlib import rcParams

# ======================
# CONFIGURACIÓN ESTÉTICA
# ======================
COLOR_SCHEME = {
    'primary': '#E53935',
    'secondary': '#EF9A9A',
    'background': '#FFFFFF',
    'text': '#000000',
    'accent': '#D32F2F',
    'highlight': '#FFCDD2'
}

LINE_COLOR_CONTRASTE = "#1E88E5"  # Azul fuerte para mayor contraste

rcParams.update({
    'axes.facecolor': COLOR_SCHEME['background'],
    'axes.edgecolor': COLOR_SCHEME['primary'],
    'axes.labelcolor': COLOR_SCHEME['text'],
    'xtick.color': COLOR_SCHEME['text'],
    'ytick.color': COLOR_SCHEME['text'],
    'grid.color': COLOR_SCHEME['secondary'],
    'figure.facecolor': COLOR_SCHEME['background']
})

st.set_page_config(
    page_title="Dominio de la Frecuencia",
    layout="wide",
    page_icon="📉"
)

# CSS Personalizado
st.markdown(f"""
<style>
    .stApp {{
        background-color: {COLOR_SCHEME['background']};
        color: {COLOR_SCHEME['text']};
    }}
    h1, h2, h3 {{
        color: {COLOR_SCHEME['primary']} !important;
        border-bottom: 2px solid {COLOR_SCHEME['secondary']};
        padding-bottom: 8px;
    }}
    .stMarkdown {{
        line-height: 1.6;
    }}
    .stSlider > div > div > div > div {{
        background: {COLOR_SCHEME['primary']} !important;
    }}
    .stButton > button {{
        background: {COLOR_SCHEME['primary']} !important;
        color: white !important;
    }}
</style>
""", unsafe_allow_html=True)

# ======================
# CONTENIDO PRINCIPAL
# ======================
st.title("Fase 1: El Dominio de la Frecuencia")
st.markdown("""
<div style='color:#000000;'>
Analiza cómo una señal puede verse tanto en el tiempo como en la frecuencia. Para ello, se crea una señal combinando varias ondas senoidales y luego se aplica un filtro paso-bajo para dejar pasar solo las frecuencias más bajas.
<br><strong>Todas las gráficas de frecuencia muestran magnitudes en decibelios (dB).</strong>
</div>
""", unsafe_allow_html=True)

# ======================
# BARRA LATERAL
# ======================
with st.sidebar:
    st.header("🔧 Parámetros de la Señal")

    st.subheader("Componente Sinusoidal 1")
    amp1 = st.slider("Amplitud 1", 0.0, 5.0, 1.0, 0.1, key='amp1')
    freq1 = st.slider("Frecuencia 1 (Hz)", 1, 100, 10, 1, key='freq1')

    st.subheader("Componente Sinusoidal 2")
    amp2 = st.slider("Amplitud 2", 0.0, 5.0, 0.5, 0.1, key='amp2')
    freq2 = st.slider("Frecuencia 2 (Hz)", 1, 100, 50, 1, key='freq2')

    st.header("🔧 Parámetros del Filtro")
    cutoff_freq = st.slider("Frecuencia de Corte (Hz)", 1, 100, 30, 1, key='cutoff')
    filter_order = st.slider("Orden del Filtro", 1, 10, 4, 1, key='order')

# ======================
# PROCESAMIENTO
# ======================
fs = 1000
t = np.arange(0, 1, 1/fs)

s1 = amp1 * np.cos(2 * np.pi * freq1 * t)
s2 = amp2 * np.cos(2 * np.pi * freq2 * t)
x = s1 + s2

nyquist = 0.5 * fs
normal_cutoff = cutoff_freq / nyquist
b, a = butter(filter_order, normal_cutoff, btype='low', analog=False)
y = lfilter(b, a, x)

def calculate_fft_db(signal, fs):
    N = len(signal)
    yf = np.fft.fft(signal)
    xf = np.fft.fftfreq(N, 1/fs)[:N//2]
    eps = np.finfo(float).eps
    magnitude_db = 20 * np.log10(2.0/N * np.abs(yf[0:N//2]) + eps)
    return xf, magnitude_db

xf_orig, yf_orig_db = calculate_fft_db(x, fs)
xf_filt, yf_filt_db = calculate_fft_db(y, fs)

# ======================
# VISUALIZACIÓN
# ======================

st.subheader("Dominio del Tiempo")
st.markdown("Señal original y su versión filtrada.")
fig_time, ax_time = plt.subplots(figsize=(10, 4))
ax_time.plot(t, x, label='Original', color=COLOR_SCHEME['primary'], alpha=0.8)
ax_time.plot(t, y, label='Filtrada', color=COLOR_SCHEME['accent'], linestyle='--')
ax_time.set_xlabel("Tiempo (s)", color=COLOR_SCHEME['text'])
ax_time.set_ylabel("Amplitud", color=COLOR_SCHEME['text'])
ax_time.legend(frameon=False)
ax_time.grid(color=LINE_COLOR_CONTRASTE, linestyle='--', alpha=0.4)
st.pyplot(fig_time)

st.subheader("Espectro de Frecuencia de la Señal Original")
st.markdown("Transformada de Fourier de la señal compuesta antes del filtrado.")
fig_orig, ax_orig = plt.subplots(figsize=(10, 4))
ax_orig.plot(xf_orig, yf_orig_db, color=COLOR_SCHEME['primary'])
ax_orig.axvline(cutoff_freq, color=LINE_COLOR_CONTRASTE, linestyle='--', label=f'Corte ({cutoff_freq} Hz)')
ax_orig.set_xlabel("Frecuencia (Hz)", color=COLOR_SCHEME['text'])
ax_orig.set_ylabel("Magnitud (dB)", color=COLOR_SCHEME['text'])
ax_orig.legend(frameon=False)
ax_orig.grid(color=LINE_COLOR_CONTRASTE, linestyle='--', alpha=0.4)
st.pyplot(fig_orig)

st.subheader("Espectro de Frecuencia de la Señal Filtrada")
st.markdown("Transformada de Fourier después del paso por el filtro paso-bajas.")
fig_filt, ax_filt = plt.subplots(figsize=(10, 4))
ax_filt.plot(xf_filt, yf_filt_db, color=COLOR_SCHEME['accent'])
ax_filt.axvline(cutoff_freq, color=LINE_COLOR_CONTRASTE, linestyle='--')
ax_filt.set_xlabel("Frecuencia (Hz)", color=COLOR_SCHEME['text'])
ax_filt.set_ylabel("Magnitud (dB)", color=COLOR_SCHEME['text'])
ax_filt.grid(color=LINE_COLOR_CONTRASTE, linestyle='--', alpha=0.4)
st.pyplot(fig_filt)

st.subheader("Respuesta del Filtro ")
st.markdown("Ganancia del filtro en dB según la frecuencia.")
w, h = freqz(b, a, worN=8000)
fig_bode, ax_bode = plt.subplots(figsize=(10, 4))
gain_db = 20 * np.log10(np.abs(h) + np.finfo(float).eps)

ax_bode.plot(0.5 * fs * w / np.pi, gain_db, color=COLOR_SCHEME['primary'])
ax_bode.axvline(cutoff_freq, color=LINE_COLOR_CONTRASTE, linestyle='--', label='Frecuencia de corte')
ax_bode.axhline(-3, color=LINE_COLOR_CONTRASTE, linestyle='--', label='-3 dB')
ax_bode.set_xlim(0, 150)
ax_bode.set_ylim(-60, 5)
ax_bode.set_xlabel("Frecuencia (Hz)", color=COLOR_SCHEME['text'])
ax_bode.set_ylabel("Ganancia (dB)", color=COLOR_SCHEME['text'])
ax_bode.legend(frameon=False)
ax_bode.grid(color=LINE_COLOR_CONTRASTE, linestyle='--', alpha=0.4)
st.pyplot(fig_bode)
```

###Señales IQ

```python
%%writefile 3_Señales_IQ.py
import streamlit as st
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import hilbert, remez
from matplotlib import rcParams

# ======================
# CONFIGURACIÓN ESTÉTICA
# ======================
COLOR_SCHEME = {
    'primary': '#E53935',
    'secondary': '#EF9A9A',
    'background': '#FFFFFF',
    'text': '#000000',
    'accent': '#D32F2F',
    'signal_i': '#1E88E5',
    'signal_q': '#E53935',
    'highlight': '#333333'
}

rcParams.update({
    'axes.facecolor': COLOR_SCHEME['background'],
    'axes.edgecolor': COLOR_SCHEME['primary'],
    'axes.labelcolor': COLOR_SCHEME['text'],
    'xtick.color': COLOR_SCHEME['text'],
    'ytick.color': COLOR_SCHEME['text'],
    'grid.color': COLOR_SCHEME['secondary'],
    'figure.facecolor': COLOR_SCHEME['background']
})

st.set_page_config(
    page_title="Señales I/Q",
    layout="centered",
    page_icon="🌐"
)

# ======================
# CSS Personalizado
# ======================
st.markdown(f"""
<style>
    .stApp {{
        background-color: {COLOR_SCHEME['background']};
        color: {COLOR_SCHEME['text']};
    }}
    h1, h2, h3 {{
        color: {COLOR_SCHEME['primary']} !important;
        border-bottom: 2px solid {COLOR_SCHEME['secondary']};
        padding-bottom: 8px;
    }}
    .stMarkdown {{
        line-height: 1.6;
    }}
    .stSlider > div > div > div > div {{
        background: {COLOR_SCHEME['primary']} !important;
    }}
    .stButton > button {{
        background: {COLOR_SCHEME['primary']} !important;
        color: white !important;
    }}
</style>
""", unsafe_allow_html=True)

# ======================
# TÍTULO Y DESCRIPCIÓN
# ======================
st.title("Fase 2: Señales I/Q")
st.markdown("""<div style='color:#000000;'>Exploración de señales analíticas: la parte real es la señal original y la parte imaginaria  es su Transformada de Hilbert.</div>""", unsafe_allow_html=True)

# ======================
# BARRA LATERAL
# ======================
with st.sidebar:
    st.header("🔧 Parámetros de la Señal Mensaje (I)")
    signal_type = st.selectbox("Tipo de Señal", ["Seno", "Coseno", "Pulso Rectangular"])
    if signal_type in ["Seno", "Coseno"]:
        freq = st.slider("Frecuencia (Hz)", 1, 50, 5, 1)
    else:
        duty_cycle = st.slider("Ciclo de trabajo", 0.1, 0.9, 0.5, 0.1)
    st.header("🔧 Filtro Hilbert")
    num_taps = st.slider("Orden del Filtro", 11, 101, 31, 2)

# ======================
# PROCESAMIENTO
# ======================
fs = 1000
t = np.arange(0, 1, 1/fs)

if signal_type == "Seno":
    i_signal = np.sin(2 * np.pi * freq * t)
elif signal_type == "Coseno":
    i_signal = np.cos(2 * np.pi * freq * t)
else:
    i_signal = np.zeros_like(t)
    i_signal[0:int(len(t)*duty_cycle)] = 1.0

analytic_signal = hilbert(i_signal)
q_signal = np.imag(analytic_signal)

# ======================
# DISEÑO EN DOS COLUMNAS
# ======================
col1, col2 = st.columns(2)

with col1:
    st.subheader("Señales en Fase (I) y Cuadratura (Q)")
    fig_iq, ax_iq = plt.subplots(figsize=(10, 5))
    ax_iq.plot(t, i_signal, label='Señal I (Fase)', color=COLOR_SCHEME['signal_i'], linewidth=2)
    ax_iq.plot(t, q_signal, label='Señal Q (Cuadratura)', color=COLOR_SCHEME['signal_q'], linestyle='--', linewidth=2)
    ax_iq.set_xlabel("Tiempo (s)", color=COLOR_SCHEME['text'])
    ax_iq.set_ylabel("Amplitud", color=COLOR_SCHEME['text'])
    ax_iq.set_xlim(0, 0.5)
    ax_iq.legend(frameon=False)
    ax_iq.grid(color=COLOR_SCHEME['secondary'], linestyle=':', alpha=0.5)
    st.pyplot(fig_iq)

    st.markdown(f"<div style='color:{COLOR_SCHEME['text']};'><strong>Nota:</strong> La señal Q está desfasada 90° respecto a la señal I, permitiendo transmitir dos flujos de datos independientes en la misma portadora.</div>", unsafe_allow_html=True)

with col2:
    # Diagrama de Polos y Ceros (mostrar siempre)
    st.subheader("Análisis del Filtro Hilbert")
    band = [0.05 * fs / 2, 0.95 * fs / 2]
    b_hilbert = remez(num_taps, band, desired=[1], type='hilbert', fs=fs)
    zeros = np.roots(b_hilbert)
    poles = np.zeros(num_taps - 1)

    fig_pz, ax_pz = plt.subplots(figsize=(6, 6))
    unit_circle = plt.Circle((0, 0), 1, fill=False, color=COLOR_SCHEME['secondary'], linestyle='--')
    ax_pz.add_patch(unit_circle)
    ax_pz.scatter(np.real(zeros), np.imag(zeros), marker='o', color=COLOR_SCHEME['primary'], label='Ceros', alpha=0.7, s=80)
    ax_pz.scatter(np.real(poles), np.imag(poles), marker='x', color=COLOR_SCHEME['accent'], label='Polos', alpha=0.7, s=100)
    ax_pz.set_title(f"Filtro Hilbert (Orden {num_taps})", color=COLOR_SCHEME['text'])
    ax_pz.set_xlabel("Parte Real", color=COLOR_SCHEME['text'])
    ax_pz.set_ylabel("Parte Imaginaria", color=COLOR_SCHEME['text'])
    ax_pz.grid(color=COLOR_SCHEME['secondary'], linestyle=':', alpha=0.5)
    ax_pz.legend(frameon=False)
    ax_pz.axis('equal')
    st.pyplot(fig_pz)

    st.markdown("""<div style='color:#000000;'>
    <strong>Interpretación:</strong><br>
    - <strong>Ceros (○):</strong> Frecuencias que el filtro atenúa<br>
    - <strong>Polos (×):</strong> Todos en el origen (z=0) para filtros FIR estables
    </div>""", unsafe_allow_html=True)

    # Espectro (mostrar siempre)
    st.subheader("Espectro de la Señal Analítica (I + jQ)")
    N = len(analytic_signal)
    yf = np.fft.fftshift(np.fft.fft(analytic_signal))
    xf = np.fft.fftshift(np.fft.fftfreq(N, 1/fs))

    fig_fft, ax_fft = plt.subplots(figsize=(10, 4))
    ax_fft.plot(xf, np.abs(yf), color=COLOR_SCHEME['primary'])
    ax_fft.set_xlabel("Frecuencia (Hz)", color=COLOR_SCHEME['text'])
    ax_fft.set_ylabel("Magnitud", color=COLOR_SCHEME['text'])
    ax_fft.grid(color=COLOR_SCHEME['secondary'], linestyle=':', alpha=0.5)
    st.pyplot(fig_fft)

    st.markdown("""<div style='color:#000000;'>
    La señal analítica ideal solo contiene componentes de frecuencia positivas.<br>
    Las pequeñas componentes negativas son artefactos de la implementación numérica.
    </div>""", unsafe_allow_html=True)
```

###Canal Ruido

```python
%%writefile 5_Canal_Ruido.py
import streamlit as st
import numpy as np
import matplotlib.pyplot as plt

# Paleta de colores minimalista en rojos
COLOR_SCHEME = {
    'primary': '#E53935',     # Rojo intenso
    'secondary': '#EF9A9A',   # Rojo claro
    'accent': '#D32F2F',      # Rojo medio
    'background': '#F5F5F5',  # Gris claro
    'text': '#333333'         # Gris oscuro
}

def dec2binarray_custom(decimal_values, num_bits):
    binary_array = np.zeros((len(decimal_values), num_bits), dtype=int)
    for i, dec_val in enumerate(decimal_values):
        binary_repr = bin(dec_val)[2:].zfill(num_bits)
        binary_array[i, :] = [int(bit) for bit in binary_repr]
    return binary_array

def qam_mod(data, M):
    if not (M > 0 and (M & (M - 1) == 0)):
        st.error("M debe ser una potencia de 2 para QAM.")
        return None

    k = int(np.log2(M))
    if len(data) % k != 0:
        padding_needed = k - (len(data) % k)
        data = np.concatenate((data, np.zeros(padding_needed, dtype=int)))

    data_reshaped = data.reshape(-1, k)
    symbols = np.zeros(data_reshaped.shape[0], dtype=complex)
    pam_levels = 2 * np.arange(np.sqrt(M)) - (np.sqrt(M) - 1)

    for i in range(data_reshaped.shape[0]):
        bits = data_reshaped[i]
        bits_i = bits[:k//2]
        bits_q = bits[k//2:]
        val_i = int("".join(str(x) for x in bits_i), 2)
        val_q = int("".join(str(x) for x in bits_q), 2)
        symbol_i = pam_levels[val_i]
        symbol_q = pam_levels[val_q]
        symbols[i] = symbol_i + 1j * symbol_q

    energy_per_symbol = np.mean(np.abs(symbols)**2)
    mod_symbols = symbols / np.sqrt(energy_per_symbol)

    return mod_symbols

def qam_demod(received_symbols, M):
    k = int(np.log2(M))
    ref_data = np.arange(M)
    ref_bits = dec2binarray_custom(ref_data, k)
    pam_levels = 2 * np.arange(np.sqrt(M)) - (np.sqrt(M) - 1)

    ref_symbols_unnorm = np.zeros(M, dtype=complex)
    for i in range(M):
        bits = ref_bits[i]
        bits_i = bits[:k//2]
        bits_q = bits[k//2:]
        val_i = int("".join(str(x) for x in bits_i), 2)
        val_q = int("".join(str(x) for x in bits_q), 2)
        symbol_i = pam_levels[val_i]
        symbol_q = pam_levels[val_q]
        ref_symbols_unnorm[i] = symbol_i + 1j * symbol_q

    energy_per_symbol_ref = np.mean(np.abs(ref_symbols_unnorm)**2)
    ref_constellation = ref_symbols_unnorm / np.sqrt(energy_per_symbol_ref)

    demod_bits = []
    for rx_symbol in received_symbols:
        distances = np.abs(rx_symbol - ref_constellation)**2
        min_dist_idx = np.argmin(distances)
        demod_bits.extend(ref_bits[min_dist_idx])

    return np.array(demod_bits)

def awgn_channel(signal, snr_db):
    snr_linear = 10**(snr_db / 10)
    signal_power = np.mean(np.abs(signal)**2)
    noise_power = signal_power / snr_linear
    noise_std = np.sqrt(noise_power / 2)
    noise = noise_std * (np.random.randn(len(signal)) + 1j * np.random.randn(len(signal)))
    received_signal = signal + noise
    return received_signal

# Configuración de la página
st.set_page_config(layout="wide")

# Estilos CSS
st.markdown(f"""
    <style>
        .stApp {{
            background-color: {COLOR_SCHEME['background']};
        }}
        h1, h2, h3, h4 {{
            color: {COLOR_SCHEME['primary']};
            font-family: 'Helvetica', sans-serif;
        }}
        .stMarkdown {{
            color: {COLOR_SCHEME['text']};
        }}
        .stSlider > div > div > div > div {{
            background-color: {COLOR_SCHEME['primary']} !important;
        }}
        .stSelectbox > div > div > div {{
            border-color: {COLOR_SCHEME['primary']} !important;
        }}
        .stButton > button {{
            background-color: {COLOR_SCHEME['primary']};
            color: white;
            border: none;
            border-radius: 4px;
            padding: 0.5rem 1rem;
        }}
        .stButton > button:hover {{
            background-color: {COLOR_SCHEME['accent']};
            color: white;
        }}
    </style>
""", unsafe_allow_html=True)

st.markdown(f"<h1 style='color:{COLOR_SCHEME['primary']}'>Simulador de Canal con Ruido AWGN y Demodulación QAM 🎛️</h1>", unsafe_allow_html=True)

st.write("""
Este dashboard se diseño para simular la transmisión de señales moduladas en QAM por medio de un canal con Ruido Blanco Gaussiano Aditivo.
Se puede visualizar cómo la constelación recibida se ve afectada por el ruido y cómo se realiza la demodulación.
""")

st.sidebar.header("🔧 Parámetros de Simulación")

qam_order = st.sidebar.selectbox("Orden de QAM (M):", [4, 16, 64, 256], index=1)
snr_db = st.sidebar.slider("SNR (dB):", min_value=-5, max_value=30, value=15)
num_symbols = st.sidebar.slider("Número de Símbolos:", min_value=100, max_value=10000, value=1000, step=100)

if st.sidebar.button("Ejecutar Simulación ✅"):
    st.markdown(f"<h2 style='color:{COLOR_SCHEME['primary']}'>Resultados de la Simulación para {qam_order}-QAM</h2>", unsafe_allow_html=True)

    k = int(np.log2(qam_order))
    data_bits = np.random.randint(0, 2, num_symbols * k)

    modulated_symbols = qam_mod(data_bits, qam_order)

    if modulated_symbols is None:
        st.stop()

    st.markdown(f"<p style='color:{COLOR_SCHEME['text']}'>Se generaron {len(modulated_symbols)} símbolos modulados en {qam_order}-QAM.</p>", unsafe_allow_html=True)

    received_symbols = awgn_channel(modulated_symbols, snr_db)
    st.markdown(f"<p style='color:{COLOR_SCHEME['text']}'>Señal transmitida a través del canal AWGN con SNR = {snr_db} dB.</p>", unsafe_allow_html=True)

    demodulated_bits = qam_demod(received_symbols, qam_order)
    st.markdown(f"<p style='color:{COLOR_SCHEME['text']}'>Símbolos demodulados a bits.</p>", unsafe_allow_html=True)

    min_len = min(len(data_bits), len(demodulated_bits))
    errors = np.sum(data_bits[:min_len] != demodulated_bits[:min_len])
    ber = errors / min_len
    st.markdown(f"<p style='color:{COLOR_SCHEME['text']}'>Tasa de Error de Bit: {ber:.4f}</p>", unsafe_allow_html=True)

    st.markdown(f"<h2 style='color:{COLOR_SCHEME['primary']}'>Visualización de la Constelación </h2>", unsafe_allow_html=True)

    fig, ax = plt.subplots(figsize=(8, 8), facecolor=COLOR_SCHEME['background'])

    ax.set_facecolor('white')
    for spine in ax.spines.values():
        spine.set_edgecolor(COLOR_SCHEME['secondary'])

    ax.scatter(received_symbols.real, received_symbols.imag,
               alpha=0.6,
               color=COLOR_SCHEME['secondary'],
               label="Símbolos Recibidos")

    ref_data = np.arange(qam_order)
    ref_bits = dec2binarray_custom(ref_data, k)
    pam_levels = 2 * np.arange(np.sqrt(qam_order)) - (np.sqrt(qam_order) - 1)
    ref_symbols_unnorm = np.zeros(qam_order, dtype=complex)
    for i in range(qam_order):
        bits = ref_bits[i]
        bits_i = bits[:k//2]
        bits_q = bits[k//2:]
        val_i = int("".join(str(x) for x in bits_i), 2)
        val_q = int("".join(str(x) for x in bits_q), 2)
        symbol_i = pam_levels[val_i]
        symbol_q = pam_levels[val_q]
        ref_symbols_unnorm[i] = symbol_i + 1j * symbol_q
    energy_per_symbol_ref = np.mean(np.abs(ref_symbols_unnorm)**2)
    ideal_constellation = ref_symbols_unnorm / np.sqrt(energy_per_symbol_ref)

    ax.scatter(ideal_constellation.real, ideal_constellation.imag,
               color=COLOR_SCHEME['primary'],
               marker='x', s=100, linewidth=2,
               label="Puntos Ideales")

    ax.set_title(f"Constelación Recibida ({qam_order}-QAM, SNR = {snr_db} dB)",
                 color=COLOR_SCHEME['text'], pad=20)
    ax.set_xlabel("Parte Real", color=COLOR_SCHEME['text'])
    ax.set_ylabel("Parte Imaginaria", color=COLOR_SCHEME['text'])

    ax.axhline(0, color=COLOR_SCHEME['secondary'], linestyle='--', linewidth=0.7)
    ax.axvline(0, color=COLOR_SCHEME['secondary'], linestyle='--', linewidth=0.7)
    ax.grid(True, color=COLOR_SCHEME['secondary'], linestyle=':', alpha=0.4)

    legend = ax.legend(frameon=True)
    frame = legend.get_frame()
    frame.set_facecolor('white')
    frame.set_edgecolor(COLOR_SCHEME['secondary'])

    ax.set_aspect('equal', adjustable='box')

    st.pyplot(fig)

    st.markdown("---")
    st.write("""
    ### Cosas a tener en cuenta:
    * **Orden de QAM (M):** Cuando se utiliza un esquema QAM de mayor orden, se incrementa la cantidad de bits transmitidos por cada símbolo. Esto mejora la eficiencia espectral, pero al mismo tiempo hace que la señal sea más vulnerable al ruido.
    * **SNR (Relación Señal/Ruido):** Un mayor valor de SNR indica que la señal tiene una potencia superior en comparación con el ruido, lo cual permite que los puntos de la constelación estén más cercanos entre sí y se reduzca la probabilidad de error en la transmisión.
    * **Constelación:** La constelación representa gráficamente los puntos de señal en el plano complejo. En el transmisor, estos puntos son definidos y exactos; sin embargo, en el receptor, el ruido provoca que se dispersen alrededor de sus posiciones ideales.
    """)
```
